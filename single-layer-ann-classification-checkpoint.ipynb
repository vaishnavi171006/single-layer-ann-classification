```python
import numpy as np

class Perceptron(object):
    
    def __init__(self, eta=0.01, epochs=50):
        self.eta = eta
        self.epochs = epochs

    def train(self, X, y):

        self.w_ = np.zeros(1 + X.shape[1])
        self.errors_ = []

        for _ in range(self.epochs):
            errors = 0
            for xi, target in zip(X, y):
                update = self.eta * (target - self.predict(xi))
                self.w_[1:] +=  update * xi
                self.w_[0] +=  update
                errors += int(update != 0.0)
            self.errors_.append(errors)
        return self

    def net_input(self, X):
        return np.dot(X, self.w_[1:]) + self.w_[0]

    def predict(self, X):
        return np.where(self.net_input(X) >= 0.0, 1, -1)
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from mlxtend.plotting import plot_decision_regions
from sklearn.linear_model import Perceptron

# Load Iris dataset
iris = load_iris()

# Select two classes and two features
X = iris.data[:100, [0, 2]]
y = iris.target[:100]

# Scale the features
scaler = StandardScaler()
X = scaler.fit_transform(X)
```

```python

import numpy as np

class AdalineGD(object):

    def __init__(self, eta=0.01, epochs=50):
        self.eta = eta
        self.epochs = epochs

    def train(self, X, y):
        self.w_ = np.zeros(1 + X.shape[1])
        self.cost_ = []

        for i in range(self.epochs):
            output = self.net_input(X)
            errors = y - output

            self.w_[1:] += self.eta * X.T.dot(errors)
            self.w_[0] += self.eta * errors.sum()

            cost = (errors ** 2).sum() / 2.0
            self.cost_.append(cost)

        return self

    def net_input(self, X):
        return np.dot(X, self.w_[1:]) + self.w_[0]

    def activation(self, X):
        return self.net_input(X)

    def predict(self, X):
        return np.where(self.activation(X) >= 0.0, 1, -1)
```

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler

# Load the Iris dataset
iris = load_iris()

# Select the first 100 samples and two features
X = iris.data[:100, [0, 2]]
y = iris.target[:100]

# Convert labels: setosa = -1, versicolor = 1
y = np.where(y == 0, -1, 1)

# Scale the features
scaler = StandardScaler()
X = scaler.fit_transform(X)

print("Dataset prepared successfully!")
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
ada = AdalineGD(eta=0.01, epochs=50)
ada.train(X, y)

print("Weights:", ada.w_)
print("Final cost:", ada.cost_[-1])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.plot(range(1, len(ada.cost_) + 1), ada.cost_, marker='o')
plt.xlabel('Epochs')
plt.ylabel('Sum-squared-error cost')
plt.title('Adaline - Learning Rate 0.01')
plt.show()
```

```python
predictions = ada.predict(X)

print("Actual labels:")
print(y[:10])

print("Predicted labels:")
print(predictions[:10])

errors = (y != predictions).sum()
print("Total misclassifications:", errors, "out of", len(y))
```

```python
from mlxtend.plotting import plot_decision_regions
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plot_decision_regions(X, y, clf=ada)

plt.xlabel('Sepal length (scaled)')
plt.ylabel('Petal length (scaled)')
plt.title('AdalineGD Decision Regions')
plt.show()
```

```python
print("Algorithm: AdalineGD")
print("Learning rate:", ada.eta)
print("Number of epochs:", ada.epochs)
print("Weights and bias:", ada.w_)
print("Final cost:", ada.cost_[-1])
print("Total misclassifications:", (y != ada.predict(X)).sum())
```

```python

import numpy as np

class AdalineSGD(object):

    def __init__(self, eta=0.01, epochs=50):
        self.eta = eta
        self.epochs = epochs

    def train(self, X, y, reinitialize_weights=True):
        if reinitialize_weights:
            self.w_ = np.zeros(1 + X.shape[1])

        self.cost_ = []

        for i in range(self.epochs):
            for xi, target in zip(X, y):
                output = self.net_input(xi)
                error = target - output

                self.w_[1:] += self.eta * xi.dot(error)
                self.w_[0] += self.eta * error

            cost = ((y - self.activation(X)) ** 2).sum() / 2.0
            self.cost_.append(cost)

        return self

    def net_input(self, X):
        return np.dot(X, self.w_[1:]) + self.w_[0]

    def activation(self, X):
        return self.net_input(X)

    def predict(self, X):
        return np.where(self.activation(X) >= 0.0, 1, -1)
```

```python
ada_sgd = AdalineSGD(eta=0.01, epochs=50)
ada_sgd.train(X, y)

print("Weights:", ada_sgd.w_)
print("Final cost:", ada_sgd.cost_[-1])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.plot(
    range(1, len(ada_sgd.cost_) + 1),
    ada_sgd.cost_,
    marker='o'
)

plt.xlabel('Epochs')
plt.ylabel('Sum-squared-error cost')
plt.title('AdalineSGD - Learning Rate 0.01')
plt.show()
```

```python
predictions = ada_sgd.predict(X)

print("Actual labels:")
print(y[:10])

print("Predicted labels:")
print(predictions[:10])

errors = (y != predictions).sum()
print("Total misclassifications:", errors, "out of", len(y))
```

```python
from mlxtend.plotting import plot_decision_regions
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plot_decision_regions(X, y, clf=ada_sgd)

plt.xlabel('Sepal length (scaled)')
plt.ylabel('Petal length (scaled)')
plt.title('AdalineSGD Decision Regions')
plt.show()
```

```python
print("Algorithm: AdalineSGD")
print("Learning rate:", ada_sgd.eta)
print("Number of epochs:", ada_sgd.epochs)
print("Weights and bias:", ada_sgd.w_)
print("Final cost:", ada_sgd.cost_[-1])
print("Total misclassifications:",
      (y != ada_sgd.predict(X)).sum())
```
